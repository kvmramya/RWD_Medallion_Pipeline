````python
import nbformat as nbf

nb = nbf.v4.new_notebook()

cells = [
    nbf.v4.new_markdown_cell("""# Real-World Data (RWD) Medallion Pipeline for Diabetic Patient Readmission Risk

## Project Architecture & Overview
This notebook implements an end-to-end **Databricks Medallion Architecture** using PySpark to process raw, semi-structured Electronic Health Records (EHR) into standardized biopharmaceutical industry metrics.

```text
 🥉 BRONZE LAYER (`bronze_ehr_raw`)
 └── Ingests and stores raw JSON EHR strings along with ingestion metadata.
       │
       ▼
 🥈 SILVER LAYER (`silver_patient_omop`)
 └── Parses JSON, cleans data, deduplicates records, and maps ICD-10 diagnosis codes to OMOP CDM concept IDs.
       │
       ▼
 🥇 GOLD LAYER (`gold_diabetic_metrics`)
 └── Calculates pre-aggregated business metrics (patient counts, readmission counts, and 30-day readmission percentages).
```"""),

    nbf.v4.new_markdown_cell("""### Step 1: Ingestion & Creation of the Bronze Layer

#### Detailed Code Explanation:
* **`SparkSession`**: The primary entry point for PySpark DataFrame and SQL functionality.
* **`raw_data`**: Simulates incoming raw EHR payloads containing JSON-formatted strings (patient IDs, gender, date of birth, diagnosis arrays, and 30-day readmission flags) alongside system ingestion timestamps.
* **`createDataFrame`**: Converts the raw Python tuple list into a distributed PySpark DataFrame named `bronze_df` with explicit column names.
* **`saveAsTable`**: Persists the DataFrame as an immutable, raw Delta table named `bronze_ehr_raw` using overwrite mode."""),

    nbf.v4.new_code_cell("""from pyspark.sql import SparkSession
from pyspark.sql.functions import col, from_json, to_timestamp, current_timestamp, expr, when
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, ArrayType

# 1. Create Mock Raw JSON EHR Data (Bronze Layer)
raw_data = [
    ('{"patient_id": "P-101", "gender": "F", "dob": "1968-05-14", "diagnoses": [{"code": "E11.9", "desc": "Type 2 Diabetes"}], "readmitted_30d": 1}', "2026-09-20 08:00:00"),
    ('{"patient_id": "P-102", "gender": "M", "dob": "1975-11-22", "diagnoses": [{"code": "I10", "desc": "Hypertension"}], "readmitted_30d": 0}', "2026-09-20 08:05:00"),
    ('{"patient_id": "P-103", "gender": "F", "dob": "1954-01-30", "diagnoses": [{"code": "E11.65", "desc": "Type 2 Diabetes with Hyperglycemia"}], "readmitted_30d": 1}', "2026-09-20 08:10:00"),
    ('{"patient_id": "P-104", "gender": "M", "dob": "1982-08-03", "diagnoses": [{"code": "E11.9", "desc": "Type 2 Diabetes"}], "readmitted_30d": 0}', "2026-09-20 08:15:00")
]

bronze_df = spark.createDataFrame(raw_data, ["raw_json_payload", "ingestion_time"])
# Write as Bronze Delta Table
bronze_df.write.format("delta").mode("overwrite").saveAsTable("bronze_ehr_raw")
print("Bronze Table Created!")"""),

    nbf.v4.new_markdown_cell("""### Step 2: Data Cleaning & OMOP CDM Standardization (Silver Layer)

#### Detailed Code Explanation:
* **`ehr_schema`**: Enforces an explicit structural schema on the unstructured JSON string to prevent runtime errors and ensure reliable parsing.
* **`from_json`**: Parses the string column `raw_json_payload` into a structured object named `parsed`.
* **Column Extracts**: Maps patient IDs (`person_source_value`), casts dates (`birth_datetime`), and unnests diagnosis codes (`icd10_code`).
* **OMOP Concept Mapping (`when` / `otherwise`)**: Converts local ICD-10 codes into standardized **OMOP CDM Concept IDs** (`201826` for Type 2 Diabetes, `320128` for Essential Hypertension).
* **Data Quality Checks**: `filter()` removes records lacking valid patient IDs, and `dropDuplicates()` deduplicates patient records.
* **`saveAsTable`**: Persists processed records into the `silver_patient_omop` Delta table."""),

    nbf.v4.new_code_cell("""# 2. Define Schema for JSON Parsing
ehr_schema = StructType([
    StructField("patient_id", StringType(), True),
    StructField("gender", StringType(), True),
    StructField("dob", StringType(), True),
    StructField("readmitted_30d", IntegerType(), True),
    StructField("diagnoses", ArrayType(
        StructType([
            StructField("code", StringType(), True),
            StructField("desc", StringType(), True)
        ])
    ), True)
])

# Process Bronze to Silver
silver_df = (
    spark.table("bronze_ehr_raw")
    .withColumn("parsed", from_json(col("raw_json_payload"), ehr_schema))
    .select(
        col("parsed.patient_id").alias("person_source_value"),
        col("parsed.gender").alias("gender_source_value"),
        to_timestamp(col("parsed.dob"), "yyyy-MM-dd").alias("birth_datetime"),
        expr("parsed.diagnoses[0].code").alias("icd10_code"),
        col("parsed.readmitted_30d").alias("is_30day_readmitted"),
        col("ingestion_time")
    )
    # Map ICD-10 to OMOP Concept ID (Standardization)
    .withColumn("omop_concept_id",
        when(col("icd10_code").startswith("E11"), 201826) # OMOP Concept ID for Type 2 Diabetes
        .otherwise(320128) # OMOP Concept ID for Essential Hypertension
    )
    .filter(col("person_source_value").isNotNull())
    .dropDuplicates(["person_source_value"])
)

silver_df.write.format("delta").mode("overwrite").saveAsTable("silver_patient_omop")
print("Silver Table Created!")"""),

    nbf.v4.new_markdown_cell("""### Step 3: Clinical Endpoint Aggregations (Gold Layer)

#### Detailed Code Explanation:
* **`groupBy("omop_concept_id")`**: Groups dataset rows by standard OMOP concept disease codes.
* **`agg(...)`**: Calculates total patient volume (`total_patients`) and total 30-day readmissions (`total_readmissions`).
* **`withColumn(...)`**: Computes the readmission rate percentage: `(total_readmissions / total_patients) * 100`.
* **`display()`**: Persists metrics to `gold_diabetic_metrics` and outputs the resulting DataFrame."""),

    nbf.v4.new_code_cell("""# 3. Gold Layer: Aggregated Clinical Metrics
gold_df = (
    spark.table("silver_patient_omop")
    .groupBy("omop_concept_id")
    .agg(
        expr("count(person_source_value)").alias("total_patients"),
        expr("sum(is_30day_readmitted)").alias("total_readmissions")
    )
    .withColumn("readmission_rate_pct", (col("total_readmissions") / col("total_patients")) * 100)
)

gold_df.write.format("delta").mode("overwrite").saveAsTable("gold_diabetic_metrics")
display(gold_df)"""),

    nbf.v4.new_markdown_cell("""### Step 4: Final Aggregated Results Table

| omop_concept_id | total_patients | total_readmissions | readmission_rate_pct | Clinical Description |
| :--- | :--- | :--- | :--- | :--- |
| **201826** | 3 | 2 | 66.67% | Type 2 Diabetes |
| **320128** | 1 | 0 | 0.0% | Essential Hypertension |""")
]

nb['cells'] = cells

with open('RWD_Medallion_Pipeline.ipynb', 'w', encoding='utf-8') as f:
    nbf.write(nb, f)

print("SUCCESS: 'RWD_Medallion_Pipeline.ipynb' created successfully!")
````
